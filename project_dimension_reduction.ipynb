**Description**

Implement it in Python functions to transform a color image into grayscale (0 to 255) and into a binarized (0 and 255) black and white image.

```python
import imageio
import matplotlib.pyplot as plt

def rgb_to_grayscale(image):
    grayscale_image = []

    for row in image:
        grayscale_row = []
        for (r, g, b) in row:
            gray_value = int(0.299 * r + 0.587 * g + 0.114 * b)
            grayscale_row.append(gray_value)
        grayscale_image.append(grayscale_row)

    return grayscale_image

def rgb_to_bw(image, threshold=128 ):
    binary_image = []

    for row in image:
        binary_row = []
        for (r, g, b) in row:
            gray= int(0.299 * r + 0.587 * g + 0.114 * b)

            # Binarize using the threshold
            if gray < threshold:
                binary_row.append(0)
            else:
                binary_row.append(255)

        binary_image.append(binary_row)

    return binary_image



rgb_image = imageio.imread('Color_Test.jpg')

gray_image = rgb_to_grayscale(rgb_image)

bw_image = rgb_to_bw(rgb_image)
```

```python
#Regular image
plt.imshow(rgb_image, cmap = plt.get_cmap(name = 'gray'))
```

```python
#GraySCale
plt.imshow(gray_image, cmap = plt.get_cmap(name = 'gray'))
```

```python
#Black and White image
plt.imshow(bw_image, cmap = plt.get_cmap(name = 'gray'))
```

```python
#Testing another threshold
bw_image = rgb_to_bw(rgb_image,145)
plt.imshow(bw_image, cmap = plt.get_cmap(name = 'gray'))
```
